Reproduzindo tabelas do relatório final (PIBIC 2025 - 2026):

1. Autoencoder PKLot, classificadores CNR (cam1 - cam9), batch de 1024, média cross-domain (CNR X PKLot) baseado na técnica

```python
filtro_tabela1 = (
    (df['batch'] == 1024) &
    (df['dataset_test'].isin(pklot)) &
    (df['dataset_encoder'] == 'PKLot') &
    (df['dataset_train'].isin(cnr))
)

agg_tabela1 = (
    df.loc[filtro_tabela1]
      .groupby(['tipo_encoder', 'tecnica_fusao', 'dataset_train'])['accuracy']
      .mean()
      .sort_values(ascending=False)
      .rename('acc_mean')
      .reset_index()
)

agg_tabela1
```

2. Autoencoder CNR, classificadores PKLot, batch de 1024, média cross-domain (PKLot x CNR) baseado na técnica

```python
filtro_tabela2 = (
    (df['batch'] == 1024) &
    (df['dataset_test'].isin(cnr)) &
    (df['dataset_encoder'] == 'CNR') &
    (df['dataset_train'].isin(pklot))
)

agg_tabela2 = (
    df.loc[filtro_tabela2]
      .groupby(['tipo_encoder', 'tecnica_fusao', 'dataset_train'])['accuracy']
      .mean()
      .sort_values(ascending=False)
      .rename('acc_mean')
      .reset_index()
)

agg_tabela2
```
